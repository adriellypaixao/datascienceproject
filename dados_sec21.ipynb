```python
import os
import subprocess
import sys

# 1. Garante que as bibliotecas necessárias estão instaladas sem quebrar
try:
  import gdown
  import pandas as pd
except ImportError:
  print('Instalando dependências necessárias...')
  subprocess.check_call(
      [sys.executable, '-m', 'pip', 'install', 'pandas', 'gdown']
  )
  import gdown
  import pandas as pd

# 2. Configurações do Google Drive
url_pasta = (
    'https://drive.google.com/drive/folders/1hhYlP-cwdMd_CFx8rgHGAeXHEkwpxbKt'
)
nome_pasta_destino = 'dados_baixados'

# Baixa a pasta apenas se ela ainda não existir na máquina
if not os.path.exists(nome_pasta_destino):
  print('Baixando a pasta do Google Drive...')
  gdown.download_folder(url_pasta, output=nome_pasta_destino, quiet=False)
else:
  print('Pasta já encontrada localmente. Pulando o download...')

# 3. Localiza o arquivo .jsonl correto dentro da pasta
caminho_arquivo = os.path.join(nome_pasta_destino, 'filmes_bruto.jsonl')

if not os.path.exists(caminho_arquivo):
  for raiz, diretorios, arquivos in os.walk(nome_pasta_destino):
    for arquivo in arquivos:
      if arquivo.endswith('.jsonl'):
        caminho_arquivo = os.path.join(raiz, arquivo)
        break

# 4. Lê o arquivo de forma segura (com limite de 5.000 linhas para evitar MemoryError)
print(f'\nLendo o arquivo de forma otimizada: {caminho_arquivo}')
df = pd.read_json(caminho_arquivo, lines=True, nrows=5000)

# 5. Executa os comandos básicos de inspeção
print('\n' + '=' * 40)
print('--- SHAPE (Linhas e Colunas) ---')
print(df.shape)

print('\n' + '=' * 40)
print('--- INFO (Tipos de dados) ---')
df.info()

print('\n' + '=' * 40)
print('--- HEAD (Primeiras linhas) ---')
display(df.head())
print('=' * 40)
```
